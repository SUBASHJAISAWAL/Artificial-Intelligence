```python
import random
import math


def calculate_cost(state):
    cost = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if state[i] == state[j]:
                cost += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost


def generate_neighbor(state):
    neighbor = state.copy()
    i, j = random.sample(range(8), 2)
    neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
    return neighbor


def simulated_annealing(initial_state):
    current = initial_state.copy()

    T = 100
    cooling_rate = 0.95
    iteration = 0

    print("Initial State:", current)
    print("Initial Cost:", calculate_cost(current))

    while T > 0.1:
        iteration += 1

        print("\n========== ITERATION", iteration, "==========")

        next_state = generate_neighbor(current)

        current_cost = calculate_cost(current)
        next_cost = calculate_cost(next_state)

        delta_E = current_cost - next_cost

        print("Current State:", current)
        print("Next State:", next_state)

        print("Current Cost:", current_cost)
        print("Next Cost:", next_cost)

        print("Delta E:", delta_E)

        if delta_E > 0:
            current = next_state
            print("Decision: Better state accepted")
            print("Probability: 1")

        else:
            probability = math.exp(delta_E / T)
            random_value = random.random()

            print("Acceptance Probability:", probability)
            print("Random Value:", random_value)

            if random_value < probability:
                current = next_state
                print("Decision: Worse state accepted")
            else:
                print("Decision: Worse state rejected")

        T = T * cooling_rate

        print("Updated Temperature:", T)

        if calculate_cost(current) == 0:
            print("\nGoal State Reached!")
            break

    return current


initial_state = [0, 1, 2, 3, 4, 5, 6, 7]

solution = simulated_annealing(initial_state)

print("\n========== FINAL RESULT ==========")
print("Initial State:", initial_state)
print("Final State:", solution)
print("Final Cost:", calculate_cost(solution))

if calculate_cost(solution) == 0:
    print("Goal State Reached Successfully!")
else:
    print("Solution Not Found")
```
