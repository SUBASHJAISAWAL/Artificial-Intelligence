```python
def dfs(state, goal, visited, path):

    if state == goal:
        return path

    visited.add(state)

    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = []

    if row > 0:
        moves.append(zero - 3)

    if row < 2:
        moves.append(zero + 3)

    if col > 0:
        moves.append(zero - 1)

    if col < 2:
        moves.append(zero + 1)

    for new_zero in moves:

        new_state = list(state)

        new_state[zero], new_state[new_zero] = \
            new_state[new_zero], new_state[zero]

        new_state = tuple(new_state)

        if new_state not in visited:

            result = dfs(
                new_state,
                goal,
                visited,
                path + [new_state]
            )

            if result is not None:
                return result

    return None


initial = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

visited = set()

solution = dfs(
    initial,
    goal,
    visited,
    [initial]
)

for state in solution:
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```

```python
def get_next_states(state):
    # Find the position of blank (0)
    zero = state.index(0)

    # Find row and column of blank
    row = zero // 3
    col = zero % 3

    moves = []

    # UP
    if row > 0:
        moves.append(zero - 3)

    # DOWN
    if row < 2:
        moves.append(zero + 3)

    # LEFT
    if col > 0:
        moves.append(zero - 1)

    # RIGHT
    if col < 2:
        moves.append(zero + 1)

    next_states = []

    # Generate new states
    for new_zero in moves:
        new_state = list(state)

        # Swap blank with adjacent tile
        new_state[zero], new_state[new_zero] = (
            new_state[new_zero],
            new_state[zero],
        )

        next_states.append(tuple(new_state))

    return next_states


def depth_limited_dfs(state, goal, depth, path):
    # Check whether goal is reached
    if state == goal:
        return path

    # Stop if depth limit is reached
    if depth == 0:
        return None

    # Generate all possible next states
    for new_state in get_next_states(state):
        # Avoid going back to a state already in current path
        if new_state not in path:
            result = depth_limited_dfs(
                new_state, goal, depth - 1, path + [new_state]
            )

            # If solution is found, return it
            if result is not None:
                return result

    # No solution from this branch
    return None


def ids(initial, goal, max_depth=50):
    depth = 0

    while depth <= max_depth:
        print("Searching with depth limit:", depth)

        result = depth_limited_dfs(initial, goal, depth, [initial])

        # Solution found
        if result is not None:
            return result

        # Increase depth limit
        depth = depth + 1

    return None


# --- INITIAL AND GOAL STATES ---

initial = (1, 2, 3, 0, 4, 6, 7, 5, 8)

goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)


# --- RUN IDS ---

solution = ids(initial, goal)


# --- PRINT SOLUTION ---

if solution is not None:
    print("\nSolution Found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for i, state in enumerate(solution):
        print("Step", i)
        print(state[0], state[1], state[2])
        print(state[3], state[4], state[5])
        print(state[6], state[7], state[8])
        print()

else:
    print("\nNo solution found within the maximum depth limit.")
```
