```python
import heapq

# Goal state
goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Manhattan Distance Heuristic
def manhattan_distance(state):
    distance = 0

    for i in range(9):

        # Do not count the blank tile
        if state[i] == 0:
            continue

        # Current position
        current_row = i // 3
        current_col = i % 3

        # Find goal position of this tile
        goal_position = goal.index(state[i])

        goal_row = goal_position // 3
        goal_col = goal_position % 3

        # Add row distance + column distance
        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# Generate possible next states
def get_neighbors(state):

    neighbors = []

    # Find blank position
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check valid position
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            # Copy current state
            new_state = list(state)

            # Move tile into blank
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors


# Print puzzle state
def print_state(state):

    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])

    print()


# A* Search
def a_star(start):

    open_list = []

    # Initial g, h and f
    g = 0
    h = manhattan_distance(start)
    f = g + h

    # Add start state to min-heap
    heapq.heappush(
        open_list,
        (f, g, start, [start])
    )

    # States already visited
    visited = set()

    while open_list:

        # Get state with minimum f
        f, g, current, path = heapq.heappop(open_list)

        # Check goal
        if current == goal:

            print("Goal reached!")
            print("Number of moves:", g)

            for state in path:
                print_state(state)

            return

        # Skip already visited state
        if current in visited:
            continue

        visited.add(current)

        # Generate neighbors
        for neighbor in get_neighbors(current):

            if neighbor not in visited:

                # Calculate new g, h and f
                new_g = g + 1
                new_h = manhattan_distance(neighbor)
                new_f = new_g + new_h

                new_path = path + [neighbor]

                # Add to min-heap
                heapq.heappush(
                    open_list,
                    (new_f, new_g, neighbor, new_path)
                )

    print("No solution found.")


# Initial state
start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)


# Run A*
a_star(start)
```
